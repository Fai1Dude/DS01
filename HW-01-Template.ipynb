## B-6

```python
# Answers to B-6
Size = ['2', '2', '3', '3', '2', '4', '4', '4', '2', '2', '2', '4', '2', '4']
from scipy import stats
import numpy as np
SizeArray = np.array(Size)
Size_labels,Size_counts = np.unique(SizeArray,return_counts=True)

def makeAutopct(Size):
    def mineAutopct(pct):
        Total = len(Size)
        val = int(round(pct*Total/100.00))
        return'{p:.1f}% ({v:d})'.format(p=pct,v=val)
    return mineAutopct

plt.figure()
plt.pie(Size_counts, labels=Size_labels, autopct=makeAutopct(Size))
plt.title("pie chart for the Size")
plt.show()
print(f'mode is{stats.mode(Size)[0]}.')
```
